```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
#celsius = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=float)
#fahrenheit = np.array([-40, 14, 32, 46, 59, 72, 100], dtype=float)
data = pd.read_csv('celsius_a_fahrenheit.csv')
celsius = np.array(data["Celsius"], dtype=float)
fahrenheit = np.array(data["Fahrenheit"], dtype=float)

data.head()
```

```python
sns.scatterplot(x='Celsius', y='Fahrenheit', data=data) 
```

```python
plt.scatter(celsius, fahrenheit)

plt.title("Correlacion entre Celsius y Fahrenheit")
plt.xlabel("Celsius")
plt.ylabel("Fahrenheit")

plt.show()
```

```python
x_train = celsius
Y_train = fahrenheit
```

```python
print("Creando el modelo")
model = tf.keras.Sequential()
model.add(tf.keras.layers.Input(shape=(1,)))
model.add(tf.keras.layers.Dense(units=1))
```

```python
print("Compilacion el modelo")
model.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss="mean_squared_error")
```

```python
print("Entrenando el modelo")
epochs_hist = model.fit(x_train, Y_train, epochs=100)
```

```python
print("Evaluando el modelo entrenado")
print("Keys: ")
print(epochs_hist.history.keys())

plt.plot(epochs_hist.history["loss"])
plt.title("Progreso de perdida durante el entrenamiento del modelo")
plt.xlabel("Epoch")
plt.ylabel("Training loss")
plt.legend(["Training Loss"])
plt.show()
```

```python
Temp_C = 100
Temp_C_array = np.array([[Temp_C]])
Temp_F = model.predict(Temp_C_array)

print("Temperatura de prediccion: " + str(Temp_F))
```
