```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
data = pd.read_csv('celsius_a_fahrenheit.csv')
celsius = np.array(data["Celsius"], dtype=float)
fahrenheit = np.array(data["Fahrenheit"], dtype=float)

data.head()
```

```python
sns.scatterplot(x='Celsius', y='Fahrenheit', data=data) 
```

```python
plt.scatter(celsius, fahrenheit)

plt.title("Correlacion entre Celsius y Fahrenheit")
plt.xlabel("Celsius")
plt.ylabel("Fahrenheit")

plt.show()
```

```python
x_train = celsius
Y_train = fahrenheit
```

```python
print("Creando el modelo")
model = tf.keras.Sequential()
model.add(tf.keras.layers.Input(shape=(1,)))
model.add(tf.keras.layers.Dense(units=1))
```

```python
print("Compilacion el modelo")
model.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss="mean_squared_error")
```

```python
print("Entrenando el modelo")
epochs_hist = model.fit(x_train, Y_train, epochs=100)
```

```python
print("Evaluando el modelo entrenado")
print("Keys: ")
print(epochs_hist.history.keys())

plt.plot(epochs_hist.history["loss"])
plt.title("Progreso de perdida durante el entrenamiento del modelo")
plt.xlabel("Epoch")
plt.ylabel("Training loss")
plt.legend(["Training Loss"])
plt.show()
```

# Preguntas

## 1
El optimizador utilizado en el modelo es Adam (Adaptive Moment Estimation). Este optimizador es uno de los más utilizados en redes neuronales porque combina dos estrategias importantes: el momentum, que ayuda a acelerar el aprendizaje en la dirección correcta del gradiente, y la adaptación automática de la tasa de aprendizaje para cada parámetro. Gracias a estas características, Adam suele converger más rápido y de forma más estable que otros optimizadores, especialmente cuando se trabaja con conjuntos de datos pequeños o con gradientes que cambian mucho durante el entrenamiento.

Otro optimizador común es SGD (Stochastic Gradient Descent). Este método actualiza los pesos del modelo utilizando directamente el gradiente calculado en cada iteración. Aunque es un algoritmo más simple y fácil de entender, generalmente requiere un ajuste cuidadoso de la tasa de aprendizaje y suele necesitar más épocas para converger. Si se hubiera utilizado SGD en este modelo, probablemente el entrenamiento habría sido más lento y podría haber requerido más iteraciones para alcanzar un error similar al obtenido con Adam.

También existe el optimizador RMSProp, que ajusta la tasa de aprendizaje dividiendo el gradiente entre una media mevil de los gradientes al cuadrado. Esto ayuda a evitar pasos demasiado grandes durante el entrenamiento y mejora la estabilidad del aprendizaje. Si se hubiera utilizado RMSProp en este modelo, es probable que el entrenamiento hubiera sido relativamente rápido y estable, aunque en muchos casos Adam suele ofrecer mejores resultados porque combina las ventajas de RMSProp con el momentum.


## 2

La función de pérdida utilizada en el modelo es mean_squared_error (MSE). Esta función mide qué tan lejos están las predicciones del modelo respecto a los valores reales. En terminos simples, calcula el promedio del cuadrado de la diferencia entre el valor real y el valor predicho. Esto es especialmente útil en problemas de regresión, como en este caso donde el modelo aprende la relación entre grados Celsius y Fahrenheit. La función MSE es importante porque guía el proceso de entrenamiento. Durante cada época, el optimizador intenta minimizar esta pérdida ajustando los pesos de la red neuronal. Al penalizar más fuertemente los errores grandes, MSE ayuda a que el modelo se acerque lo más posible a los valores reales.

Otra función de pérdida que se podría utilizar es Mean Absolute Error (MAE). Esta función calcula el promedio del valor absoluto de los errores entre las predicciones y los valores reales. A diferencia de MSE, MAE no penaliza tanto los errores grandes, lo que puede hacer que el modelo sea más robusto frente a valores atípicos. Si se utilizara MAE en este modelo, el entrenamiento podría ser un poco más estable frente a errores grandes, aunque en este caso específico, donde los datos son bastante limpios y siguen una relación lineal clara, MSE suele proporcionar un ajuste más preciso.

## 4

Las epochs representan el número de veces que el modelo recorre todo el conjunto de datos de entrenamiento. Son fundamentales porque permiten que el modelo aprenda gradualmente los patrones presentes en los datos. En cada época, el modelo ajusta sus pesos con base en el error obtenido en las predicciones.

Si se utilizan muy pocas epochs, el modelo puede quedar subentrenado, lo que significa que no ha tenido suficientes oportunidades para aprender la relación entre las variables. Por otro lado, si se utilizan demasiadas epochs, el modelo podría sobreajustarse (overfitting) o continuar entrenando sin obtener mejoras significativas. Por lo tanto, encontrar un número adecuado de epochs es importante para lograr un buen equilibrio entre precisión y eficiencia.


## 5

En el modelo original no se especifica explícitamente una función de activación en las capas Dense. Cuando esto ocurre en TensorFlow/Keras, se utiliza por defecto la función lineal (linear). Esto significa que la salida de la capa es simplemente una combinación lineal de los pesos y las entradas.

Sí es posible cambiar la función de activación. Una opción común es utilizar ReLU (Rectified Linear Unit).

Esta función devuelve cero cuando el valor es negativo y deja pasar los valores positivos. Es muy utilizada en redes neuronales porque ayuda a mejorar la eficiencia del entrenamiento y reduce el problema del desvanecimiento del gradiente.
