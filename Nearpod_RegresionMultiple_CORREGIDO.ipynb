# Regresión lineal: Habitaciones ocupadas vs Ingreso del restaurante
Notebook corregido con:
- Cálculo correcto de correlación de Pearson
- Regresión lineal
- Cálculo de R² (varianza explicada)
- Interpretación lista para entregar

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Cargar datos
df = pd.read_csv('datos_hotel.csv')
df.head()
```

```python
# Gráfica de dispersión
plt.figure()
plt.scatter(df['Habitaciones_ocupadas'], df['Ingreso'])
plt.xlabel('Habitaciones ocupadas')
plt.ylabel('Ingreso')
plt.title('Relación entre habitaciones ocupadas e ingreso')
plt.show()
```

```python
# Regresión lineal
X = df[['Habitaciones_ocupadas']]
y = df['Ingreso']

model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

pendiente = model.coef_[0]
interseccion = model.intercept_
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print('Pendiente:', pendiente)
print('Intersección:', interseccion)
print('Error cuadrático medio:', mse)
print('Varianza explicada (R²):', r2)
```

```python
# Correlación de Pearson (correcta)
corr = df['Habitaciones_ocupadas'].corr(df['Ingreso'])
print('Coeficiente de correlación de Pearson:', corr)
```

```python
# Gráfica con recta de regresión
plt.figure()
plt.scatter(df['Habitaciones_ocupadas'], df['Ingreso'], label='Datos')
plt.plot(df['Habitaciones_ocupadas'], y_pred, label='Regresión')
plt.xlabel('Habitaciones ocupadas')
plt.ylabel('Ingreso')
plt.title('Regresión lineal')
plt.legend()
plt.show()
```

## Conclusión
La gráfica de dispersión muestra una relación positiva entre el número de habitaciones ocupadas y el ingreso diario del restaurante, ya que la línea de tendencia presenta una pendiente ascendente. Esto indica que, en general, cuando aumenta la ocupación del hotel, el ingreso del restaurante también tiende a incrementarse, debido a que un mayor número de huéspedes representa más clientes potenciales.

Sin embargo, el coeficiente de determinación obtenido (R² ≈ 0.18) indica que el número de habitaciones ocupadas explica aproximadamente el 18% de la variación del ingreso del restaurante. Esto significa que la relación es positiva pero moderada-débil, ya que el ingreso también depende de otros factores como el consumo promedio por huésped, promociones, eventos o el día analizado.
