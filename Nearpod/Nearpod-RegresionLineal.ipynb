```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
%matplotlib inline

#Librerias para operaciones de regresion lineal 
from sklearn import linear_model 
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = pd.read_csv("./datos_hotel.csv")
```

```python
#Determinen la cantidad de filas y columnas del archivo
data.shape
```

```python
#Mostrar la tabla 
data.head()
```

```python
#Seleccionar los datos para determinar el coeficiente de correlacion
X_train = np.array(data["Habitaciones_ocupadas"]).reshape(-1,1)
y_train = np.array(data["Ingreso"])
```

```python
#Calcular la formula de regresion lineal , coeficiente de correlación lineal
r = linear_model.LinearRegression()

#Entrenamos al modelo con los datos del entrenamiento
r.fit(X_train, y_train)

#obtener predicciones
y_pred = r.predict(X_train)

print('Coeficiente de correlación: ' , r.coef_)
print('Interseccion: ' , r.intercept_)
print('Error cuadrático de la media: ', mean_squared_error(y_train, y_pred))
print('Varianza: ' , r2_score(y_train, y_pred))
```

```python
x = X_train.ravel()
plt.figure(figsize=(8,5))
plt.scatter(x,y_train,color='steelblue', s=80, label = 'Datos')
plt.plot(x,y_pred, color = 'red',linewidth=3,label='Pendiente')
plt.xlabel('Habitaciones Ocupadas')
plt.ylabel('Ingreso')
plt.title('Nearpod')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

La gráfica de dispersión muestra una relación positiva entre el número de habitaciones ocupadas y el ingreso diario del restaurante. La línea de tendencia tiene pendiente ascendente, lo que indica que, en general, cuando aumenta la ocupación del hotel, también tiende a aumentar el ingreso del restaurante. Esto es lógico, ya que un mayor número de huéspedes representa más clientes potenciales que pueden consumir en el restaurante.

Sin embargo, los puntos presentan cierta dispersión, lo que indica que la relación no es perfecta y que el ingreso también depende de otros factores. El coeficiente de correlación obtenido (r ≈ 0.47) confirma que existe una correlación positiva moderada, es decir, las variables están relacionadas de forma directa, pero la ocupación no explica completamente el comportamiento del ingreso del restaurante.
